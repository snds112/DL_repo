```python
import time
import matplotlib.pyplot as plt
from tensorflow.keras.datasets import imdb
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, Conv1D, Dense, LSTM, GlobalMaxPooling1D
```

```python
num_words = 10000
maxlen = 200

(x_train, y_train), (x_test, y_test) = imdb.load_data(num_words=num_words)

x_train = pad_sequences(x_train, maxlen=maxlen)
x_test = pad_sequences(x_test, maxlen=maxlen)
```

```python
model_conv = Sequential()
model_conv.add(Embedding(input_dim=10000, output_dim=32, input_length=maxlen))

# Convolution layer
model_conv.add(Conv1D(filters=64, kernel_size=5, activation='relu'))

# Pooling
model_conv.add(GlobalMaxPooling1D())

# Output
model_conv.add(Dense(1, activation='sigmoid'))

model_conv.compile(
    optimizer='adam',
    loss='binary_crossentropy',
    metrics=['accuracy']
)

start_time = time.time()

history_conv = model_conv.fit(
    x_train, y_train,
    epochs=5,
    batch_size=32,
    validation_split=0.2
)

time_conv = time.time() - start_time

loss_conv, acc_conv = model_conv.evaluate(x_test, y_test)

print("Conv1D test accuracy:", acc_conv)
print("Conv1D training time:", time_conv)
```

```python
print("\nTraining Conv1D (kernel_size=3):")

model_conv_k3 = Sequential()
model_conv_k3.add(Embedding(10000, 32, input_length=maxlen))
model_conv_k3.add(Conv1D(64, 3, activation='relu'))
model_conv_k3.add(GlobalMaxPooling1D())
model_conv_k3.add(Dense(1, activation='sigmoid'))

model_conv_k3.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])

history_conv_k3 = model_conv_k3.fit(
    x_train, y_train,
    epochs=5,
    batch_size=32,
    validation_split=0.2
)

loss_k3, acc_k3 = model_conv_k3.evaluate(x_test, y_test)

print("Conv1D (kernel=3) accuracy:", acc_k3)
```

```python
print("\nTraining Conv1D (kernel_size=7)...")

model_conv_k7 = Sequential()
model_conv_k7.add(Embedding(10000, 32, input_length=maxlen))
model_conv_k7.add(Conv1D(64, 7, activation='relu'))
model_conv_k7.add(GlobalMaxPooling1D())
model_conv_k7.add(Dense(1, activation='sigmoid'))

model_conv_k7.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])

history_conv_k7 = model_conv_k7.fit(
    x_train, y_train,
    epochs=5,
    batch_size=32,
    validation_split=0.2
)

loss_k7, acc_k7 = model_conv_k7.evaluate(x_test, y_test)

print("Conv1D (kernel=7) accuracy:", acc_k7)
```

```python
print("\nTraining Conv1D (kernel_size=3 and filters=128)...")

model_conv_filters_128 = Sequential()
model_conv_filters_128.add(Embedding(10000, 32, input_length=maxlen))
model_conv_filters_128.add(Conv1D(256, 3, activation='relu'))
model_conv_filters_128.add(GlobalMaxPooling1D())
model_conv_filters_128.add(Dense(1, activation='sigmoid'))

model_conv_filters_128.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])

start_time = time.time()

history_conv_filters_128 = model_conv_filters_128.fit(
    x_train, y_train,
    epochs=5,
    batch_size=32,
    validation_split=0.2
)

end_time = time.time() - start_time

loss_filters_128, acc_filters_128 = model_conv_filters_128.evaluate(x_test, y_test)

print("Conv1D (kernel=3 and filters=128) accuracy:", acc_filters_128)
print("Conv1D (kernel=3 and filters=128) training time:", end_time)
```

```python
print("\nTraining LSTM model...")

model_lstm = Sequential()
model_lstm.add(Embedding(input_dim=10000, output_dim=32, input_length=maxlen))
model_lstm.add(LSTM(64))
model_lstm.add(Dense(1, activation='sigmoid'))

model_lstm.compile(
    optimizer='adam',
    loss='binary_crossentropy',
    metrics=['accuracy']
)

start_time = time.time()

history_lstm = model_lstm.fit(
    x_train, y_train,
    epochs=5,
    batch_size=32,
    validation_split=0.2
)

time_lstm = time.time() - start_time

loss_lstm, acc_lstm = model_lstm.evaluate(x_test, y_test)

print("LSTM Test Accuracy:", acc_lstm)
print("LSTM Training Time:", time_lstm)
```

```python
plt.figure()

plt.plot(history_conv.history['val_accuracy'], label='Conv1D (k=5)')
plt.plot(history_lstm.history['val_accuracy'], label='LSTM')

plt.title("Conv1D vs LSTM")
plt.xlabel("Epochs")
plt.ylabel("Validation accuracy")
plt.legend()

plt.show()
```
