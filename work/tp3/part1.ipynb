# EXO1:

```python
import tensorflow as tf
from tensorflow.keras import layers, models

# 1. Load and normalize the data
(train_images, train_labels), (test_images, test_labels) = tf.keras.datasets.mnist.load_data()

# Reshape to (28, 28, 1) and normalize pixel values to be between 0 and 1
train_images = train_images.reshape((60000, 28, 28, 1)).astype('float32') / 255
test_images = test_images.reshape((10000, 28, 28, 1)).astype('float32') / 255

# 2 & 3. Build the model architecture
model = models.Sequential([
    # First convolutional layer (32 filters, 3x3 kernel) + Max-pooling
    layers.Conv2D(32, (3, 3), activation='relu', input_shape=(28, 28, 1)),
    layers.MaxPooling2D((2, 2)),

    # Second convolutional layer (64 filters, 3x3 kernel) + Max-pooling
    layers.Conv2D(64, (3, 3), activation='relu'),
    layers.MaxPooling2D((2, 2)),

    # Flatten and Dense layers
    layers.Flatten(),
    layers.Dense(128, activation='relu'),
    layers.Dense(10, activation='softmax')
])

# 4. Compile the model
model.compile(optimizer='adam',
              loss='sparse_categorical_crossentropy',
              metrics=['accuracy'])

# 5. Train for 10 epochs and evaluate
model.fit(train_images, train_labels, epochs=10, batch_size=64)

test_loss, test_acc = model.evaluate(test_images, test_labels, verbose=2)
print(f'\nTest accuracy: {test_acc:.4f}')
```

# EXO2:

```python
import tensorflow as tf
from tensorflow.keras import layers, models
import matplotlib.pyplot as plt

# 1. Load Fashion-MNIST
print("Loading Fashion-MNIST data...")
fashion_mnist = tf.keras.datasets.fashion_mnist
(train_images, train_labels), (test_images, test_labels) = fashion_mnist.load_data()

# Normalize pixel values to be between 0 and 1
train_images, test_images = train_images / 255.0, test_images / 255.0

# Reshape for CNN (needs the color channel dimension)
train_images_cnn = train_images.reshape((60000, 28, 28, 1))
test_images_cnn = test_images.reshape((10000, 28, 28, 1))

# --- TASK 1: Simple ANN (One hidden layer, 256 neurons) ---
print("\n--- Building Simple ANN ---")
ann_model = models.Sequential([
    layers.Flatten(input_shape=(28, 28)), # Direct flatten!
    layers.Dense(256, activation='relu'),
    layers.Dense(10, activation='softmax')
])

ann_model.compile(optimizer='adam',
                  loss='sparse_categorical_crossentropy',
                  metrics=['accuracy'])

print("\nANN Summary:")
ann_model.summary()

print("\nTraining ANN...")
ann_history = ann_model.fit(train_images, train_labels, epochs=10,
                            validation_data=(test_images, test_labels), verbose=1)

# --- TASK 2: CNN (Similar to Exercise 1) ---
print("\n--- Building CNN ---")
cnn_model = models.Sequential([
    layers.Conv2D(32, (3, 3), activation='relu', input_shape=(28, 28, 1)),
    layers.MaxPooling2D((2, 2)),
    layers.Conv2D(64, (3, 3), activation='relu'),
    layers.MaxPooling2D((2, 2)),
    layers.Flatten(), # The bridge we discussed!
    layers.Dense(128, activation='relu'),
    layers.Dense(10, activation='softmax')
])

cnn_model.compile(optimizer='adam',
                  loss='sparse_categorical_crossentropy',
                  metrics=['accuracy'])

print("\nCNN Summary:")
cnn_model.summary()

print("\nTraining CNN...")
cnn_history = cnn_model.fit(train_images_cnn, train_labels, epochs=10,
                            validation_data=(test_images_cnn, test_labels), verbose=1)

# --- TASK 3: Comparison Visualization ---
plt.figure(figsize=(12, 4))

plt.subplot(1, 2, 1)
plt.plot(ann_history.history['val_accuracy'], label='ANN Val Accuracy')
plt.plot(cnn_history.history['val_accuracy'], label='CNN Val Accuracy')
plt.title('Accuracy Comparison')
plt.xlabel('Epoch')
plt.ylabel('Accuracy')
plt.legend()

plt.show()
```

CNNs out perform ANNs on images because they conserve spatial detail (which features are where and where are they to eachother) while ANNs just flatten the image and lose all spatial information, they cannot detect features if they move or rotate unless the entire weight matrix is changed along with the data.

```python
import tensorflow as tf
import numpy as np
import matplotlib.pyplot as plt

# 1. Load and Normalize CIFAR-10
(x_train, _), (_, _) = tf.keras.datasets.cifar10.load_data()
img = x_train[0:8] / 255.0  # Just 8 image for clear comparison

# Define the experiments
configs = [
    ('valid', 1, "Valid / Stride 1"),
    ('same', 1, "Same / Stride 1"),
    ('valid', 2, "Valid / Stride 2"),
    ('same', 2, "Same / Stride 2")
]

plt.figure(figsize=(18, 5))

# Show Original
plt.subplot(1, 5, 1)
plt.imshow(img[0])
plt.title(f"Original\n{img.shape[1:3]}")
plt.axis('off')

# Run experiments and plot
for i, (p, s, title) in enumerate(configs):
    # Define layer
    layer = tf.keras.layers.Conv2D(filters=1, kernel_size=(3,3), strides=s, padding=p)

    # Pass image through
    output = layer(img)
    feat_map = output[0, :, :, 0] # Remove batch and channel dims

    # Plot
    plt.subplot(1, 5, i + 2)
    plt.imshow(feat_map, cmap='gray')
    plt.title(f"{title}\nShape: {feat_map.shape}")
    plt.axis('off')

plt.tight_layout()
plt.show()
```

the bigger the stride the less detail there is after the convolution, the padding adds some dark pixels at the right and bottom of the image.

# EXO4:

```python
import tensorflow as tf
from tensorflow.keras import layers, models
import matplotlib.pyplot as plt

# Load and normalize data
(train_images, train_labels), (test_images, test_labels) = tf.keras.datasets.cifar10.load_data()
train_images, test_images = train_images / 255.0, test_images / 255.0

def build_model(use_augmentation=False):
    model = models.Sequential()

    # Task 3: Add Augmentation Layers at the beginning
    if use_augmentation:
        model.add(layers.RandomFlip("horizontal", input_shape=(32, 32, 3)))
        model.add(layers.RandomRotation(0.1))

    # Task 1: 3 Convolutional Blocks
    model.add(layers.Conv2D(32, (3, 3), activation='relu', input_shape=(32, 32, 3) if not use_augmentation else None))
    model.add(layers.MaxPooling2D((2, 2)))

    model.add(layers.Conv2D(64, (3, 3), activation='relu'))
    model.add(layers.MaxPooling2D((2, 2)))

    model.add(layers.Conv2D(64, (3, 3), activation='relu'))

    model.add(layers.Flatten())
    model.add(layers.Dense(64, activation='relu'))
    model.add(layers.Dense(10))

    model.compile(optimizer='adam',
                  loss=tf.keras.losses.SparseCategoricalCrossentropy(from_logits=True),
                  metrics=['accuracy'])
    return model




# Train Baseline
print("\n--- Training Baseline Model ---")
baseline_model = build_model(use_augmentation=False)
baseline_history = baseline_model.fit(train_images, train_labels, epochs=30,
                                    validation_data=(test_images, test_labels))

# Train Augmented
print("\n--- Training Augmented Model ---")
aug_model = build_model(use_augmentation=True)
aug_history = aug_model.fit(train_images, train_labels, epochs=30,
                            validation_data=(test_images, test_labels))
```

```python
plt.figure(figsize=(12, 5))

# Baseline Plot
plt.subplot(1, 2, 1)
plt.plot(baseline_history.history['accuracy'], label='Train Acc (Baseline)')
plt.plot(baseline_history.history['val_accuracy'], label='Val Acc (Baseline)')
plt.title('Baseline: Clear Overfitting')
plt.legend()

# Augmented Plot
plt.subplot(1, 2, 2)
plt.plot(aug_history.history['accuracy'], label='Train Acc (Augmented)')
plt.plot(aug_history.history['val_accuracy'], label='Val Acc (Augmented)')
plt.title('Augmented: Better Generalization')
plt.legend()

plt.show()
```

```python
import tensorflow as tf
from tensorflow.keras import layers, Input
from tensorflow.keras.models import Model
import matplotlib.pyplot as plt

# 1. Load and normalize data
(train_images, train_labels), (test_images, test_labels) = tf.keras.datasets.mnist.load_data()
train_images = train_images.reshape((-1, 28, 28, 1)).astype('float32') / 255
test_images = test_images.reshape((-1, 28, 28, 1)).astype('float32') / 255

# 2. Build the Model using the Functional API (The Modern Way)
inputs = Input(shape=(28, 28, 1))

# We name this layer 'first_conv' to easily extract it later
x = layers.Conv2D(32, (3, 3), activation='relu', name='first_conv')(inputs)
x = layers.MaxPooling2D((2, 2))(x)
x = layers.Conv2D(64, (3, 3), activation='relu')(x)
x = layers.MaxPooling2D((2, 2))(x)
x = layers.Flatten()(x)
x = layers.Dense(128, activation='relu')(x)
outputs = layers.Dense(10, activation='softmax')(x)

model = Model(inputs=inputs, outputs=outputs)

# 3. Compile and Train
model.compile(optimizer='adam', loss='sparse_categorical_crossentropy', metrics=['accuracy'])
model.fit(train_images, train_labels, epochs=3, batch_size=64)

# --- EXERCISE 5: FEATURE MAP VISUALIZATION ---

# 4. Define the visualization model (Task 2)
# We map the model's existing input to the output of our named layer
feature_map_model = Model(inputs=model.input, outputs=model.get_layer('first_conv').output)
```

```python
# --- EXERCISE 5: Task 7 (Comparison) ---

# Choose two different digits to compare (e.g., indices 0 and 1)
indices_to_compare = [0,1]

for idx in indices_to_compare:
    # 1. Select the image and get the actual label for the title
    sample_img = test_images[idx:idx+1]
    actual_label = test_labels[idx]

    # 2. Generate the feature maps
    feature_maps = feature_map_model.predict(sample_img)

    # 3. Plot the results
    plt.figure(figsize=(10, 10))
    plt.suptitle(f"Feature Maps for Digit: {actual_label} (Index {idx})", fontsize=16)

    for i in range(32):
        plt.subplot(8, 4, i + 1)
        plt.imshow(feature_maps[0, :, :, i], cmap='viridis')
        plt.axis('off')

    plt.tight_layout()
    plt.subplots_adjust(top=0.92)
    plt.show()
```

```python
import tensorflow as tf
from tensorflow.keras import layers, models, callbacks

# 1. Load CIFAR-10 Data
(train_images, train_labels), (test_images, test_labels) = tf.keras.datasets.cifar10.load_data()
train_images, test_images = train_images / 255.0, test_images / 255.0

# 2. Rebuild the Exercise 4 model (without augmentation)
model = models.Sequential([
    layers.Input(shape=(32, 32, 3)),
    layers.Conv2D(32, (3, 3), activation='relu'),
    layers.MaxPooling2D((2, 2)),
    layers.Conv2D(64, (3, 3), activation='relu'),
    layers.MaxPooling2D((2, 2)),
    layers.Conv2D(64, (3, 3), activation='relu'),
    layers.Flatten(),
    layers.Dense(64, activation='relu'),
    layers.Dense(10)
])

model.compile(optimizer='adam',
              loss=tf.keras.losses.SparseCategoricalCrossentropy(from_logits=True),
              metrics=['accuracy'])

# 3. Implement Callbacks (Task 2 & 3)

# EarlyStopping: Stop if validation loss doesn't improve for 3 straight epochs
early_stop = callbacks.EarlyStopping(
    monitor='val_loss',
    patience=3,
    restore_best_weights=True
)

# ModelCheckpoint: Save only the best version of the model to a file
checkpoint = callbacks.ModelCheckpoint(
    filepath='best_model.keras',
    monitor='val_loss',
    save_best_only=True
)

# 4. Train for 50 epochs (Task 4)
history = model.fit(
    train_images, train_labels,
    epochs=50,
    validation_data=(test_images, test_labels),
    callbacks=[early_stop, checkpoint] # Plug them in here
)
```

# EXO7:

```python
import tensorflow as tf
from tensorflow.keras import layers, models
import matplotlib.pyplot as plt

# 1. Load CIFAR-10
(train_images, train_labels), (test_images, test_labels) = tf.keras.datasets.cifar10.load_data()
train_images, test_images = train_images / 255.0, test_images / 255.0

# Helper function to create the models (Task 1)
def create_model(use_batchnorm=False):
    model = models.Sequential()
    model.add(layers.Input(shape=(32, 32, 3)))

    # Layer 1
    model.add(layers.Conv2D(32, (3, 3), padding='same'))
    if use_batchnorm: model.add(layers.BatchNormalization())
    model.add(layers.Activation('relu'))
    model.add(layers.MaxPooling2D((2, 2)))

    # Layer 2
    model.add(layers.Conv2D(64, (3, 3), padding='same'))
    if use_batchnorm: model.add(layers.BatchNormalization())
    model.add(layers.Activation('relu'))
    model.add(layers.MaxPooling2D((2, 2)))

    model.add(layers.Flatten())
    model.add(layers.Dense(64, activation='relu'))
    model.add(layers.Dense(10))

    model.compile(optimizer='adam',
                  loss=tf.keras.losses.SparseCategoricalCrossentropy(from_logits=True),
                  metrics=['accuracy'])
    return model

# Create both versions
model_no_bn = create_model(use_batchnorm=False)
model_with_bn = create_model(use_batchnorm=True)

# 2. Train both for 20 epochs (Task 2)
print("Training model WITHOUT Batch Normalization...")
history_no_bn = model_no_bn.fit(train_images, train_labels, epochs=20,
                                validation_data=(test_images, test_labels), verbose=0)

print("Training model WITH Batch Normalization...")
history_with_bn = model_with_bn.fit(train_images, train_labels, epochs=20,
                                 validation_data=(test_images, test_labels), verbose=0)
```

```python
plt.figure(figsize=(12, 5))

# Plot Accuracy
plt.subplot(1, 2, 1)
plt.plot(history_no_bn.history['val_accuracy'], label='No BatchNorm', linestyle='--')
plt.plot(history_with_bn.history['val_accuracy'], label='With BatchNorm')
plt.title('Validation Accuracy Comparison')
plt.xlabel('Epochs')
plt.ylabel('Accuracy')
plt.legend()

# Plot Loss
plt.subplot(1, 2, 2)
plt.plot(history_no_bn.history['val_loss'], label='No BatchNorm', linestyle='--')
plt.plot(history_with_bn.history['val_loss'], label='With BatchNorm')
plt.title('Validation Loss Comparison')
plt.xlabel('Epochs')
plt.ylabel('Loss')
plt.legend()

plt.show()
```

Over-parameterization: For a simple network, BatchNorm adds complexity that can lead to faster overfitting on the training data.

Batch Sensitivity: BatchNorm requires large batch sizes (like 128 or 256) to accurately calculate the mean and variance; otherwise, validation performance suffers.

Convergence vs. Final Accuracy: the BatchNorm model starts faster in the first few epochs (faster convergence), but it hit a lower ceiling because it wasn't tuned for such a small model.

```python
import tensorflow as tf
from tensorflow.keras import layers, models
import matplotlib.pyplot as plt

# 1. Load Fashion-MNIST
(train_images, train_labels), (test_images, test_labels) = tf.keras.datasets.fashion_mnist.load_data()
train_images = train_images.reshape((-1, 28, 28, 1)).astype('float32') / 255
test_images = test_images.reshape((-1, 28, 28, 1)).astype('float32') / 255

def build_fashion_model(use_dropout=False):
    model = models.Sequential()
    model.add(layers.Input(shape=(28, 28, 1)))
    model.add(layers.Conv2D(32, (3, 3), activation='relu'))
    model.add(layers.MaxPooling2D((2, 2)))
    model.add(layers.Flatten())
    model.add(layers.Dense(128, activation='relu'))

    if use_dropout:
        model.add(layers.Dropout(0.5)) # Task 1: 50% Dropout

    model.add(layers.Dense(10, activation='softmax'))

    model.compile(optimizer='adam',
                  loss='sparse_categorical_crossentropy',
                  metrics=['accuracy'])
    return model

# Train both models
print("Training model WITHOUT Dropout...")
model_no_dropout = build_fashion_model(use_dropout=False)
history_no = model_no_dropout.fit(train_images, train_labels, epochs=15,
                                  validation_data=(test_images, test_labels), verbose=0)

print("Training model WITH Dropout...")
model_dropout = build_fashion_model(use_dropout=True)
history_yes = model_dropout.fit(train_images, train_labels, epochs=15,
                                 validation_data=(test_images, test_labels), verbose=0)
```

```python
plt.figure(figsize=(12, 5))

# Plot for No Dropout
plt.subplot(1, 2, 1)
plt.plot(history_no.history['accuracy'], label='Train Acc', color='blue')
plt.plot(history_no.history['val_accuracy'], label='Val Acc', color='orange')
plt.title('No Dropout: The Overfitting Gap')
plt.legend()

# Plot for Dropout
plt.subplot(1, 2, 2)
plt.plot(history_yes.history['accuracy'], label='Train Acc', color='blue')
plt.plot(history_yes.history['val_accuracy'], label='Val Acc', color='orange')
plt.title('With Dropout: Generalization')
plt.legend()

plt.show()
```

```python
import tensorflow as tf
from tensorflow.keras import layers, models, callbacks
import matplotlib.pyplot as plt

# 1. Load CIFAR-10
(train_images, train_labels), (test_images, test_labels) = tf.keras.datasets.cifar10.load_data()
train_images, test_images = train_images / 255.0, test_images / 255.0

def build_model():
    model = models.Sequential([
        layers.Input(shape=(32, 32, 3)),
        layers.Conv2D(32, (3, 3), activation='relu', padding='same'),
        layers.MaxPooling2D((2, 2)),
        layers.Conv2D(64, (3, 3), activation='relu', padding='same'),
        layers.MaxPooling2D((2, 2)),
        layers.Flatten(),
        layers.Dense(64, activation='relu'),
        layers.Dense(10)
    ])
    return model

# --- Task 1: Fixed Learning Rate ---
model_fixed = build_model()
model_fixed.compile(optimizer=tf.keras.optimizers.Adam(learning_rate=0.001),
                    loss=tf.keras.losses.SparseCategoricalCrossentropy(from_logits=True),
                    metrics=['accuracy'])

print("Training with Fixed LR...")
history_fixed = model_fixed.fit(train_images, train_labels, epochs=20,
                                validation_data=(test_images, test_labels), verbose=0)

# --- Task 2: LR Scheduler ---
model_sched = build_model()
model_sched.compile(optimizer=tf.keras.optimizers.Adam(learning_rate=0.001),
                    loss=tf.keras.losses.SparseCategoricalCrossentropy(from_logits=True),
                    metrics=['accuracy'])

# Reduce LR by factor of 0.2 if val_loss doesn't improve for 3 epochs
lr_scheduler = callbacks.ReduceLROnPlateau(monitor='val_loss', factor=0.2, patience=3, min_lr=1e-6)

print("Training with LR Scheduler...")
history_sched = model_sched.fit(train_images, train_labels, epochs=20,
                                validation_data=(test_images, test_labels),
                                callbacks=[lr_scheduler], verbose=0)
```

```python
plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plt.plot(history_fixed.history['val_loss'], label='Fixed LR (0.001)', linestyle='--')
plt.plot(history_sched.history['val_loss'], label='Scheduled LR')
plt.title('Validation Loss')
plt.legend()

plt.subplot(1, 2, 2)
plt.plot(history_fixed.history['val_accuracy'], label='Fixed LR (0.001)', linestyle='--')
plt.plot(history_sched.history['val_accuracy'], label='Scheduled LR')
plt.title('Validation Accuracy')
plt.legend()

plt.show()
```

fixed LR is more "bouncy"

```python
import tensorflow as tf
from tensorflow.keras import layers, models
import numpy as np

# 1. Load CIFAR-10
(train_images, train_labels), (test_images, test_labels) = tf.keras.datasets.cifar10.load_data()
train_images, test_images = train_images / 255.0, test_images / 255.0

# Helper to build variations (Task 1)
def build_variant(filters):
    model = models.Sequential([
        layers.Input(shape=(32, 32, 3)),
        layers.Conv2D(filters, (3, 3), activation='relu'),
        layers.MaxPooling2D((2, 2)),
        layers.Conv2D(filters * 2, (3, 3), activation='relu'),
        layers.Flatten(),
        layers.Dense(64, activation='relu'),
        layers.Dense(10, activation='softmax') # Softmax is required for averaging
    ])
    model.compile(optimizer='adam', loss='sparse_categorical_crossentropy', metrics=['accuracy'])
    return model

# Train 3 models with different filter counts
print("Training Ensemble Model 1 (16 filters)...")
m1 = build_variant(16)
m1.fit(train_images, train_labels, epochs=10, verbose=0)

print("Training Ensemble Model 2 (32 filters)...")
m2 = build_variant(32)
m2.fit(train_images, train_labels, epochs=10, verbose=0)

print("Training Ensemble Model 3 (64 filters)...")
m3 = build_variant(64)
m3.fit(train_images, train_labels, epochs=10, verbose=0)

# --- Task 2: Average Predictions ---

# Get probability distributions from each model
pred1 = m1.predict(test_images)
pred2 = m2.predict(test_images)
pred3 = m3.predict(test_images)

# Average the softmax outputs
ensemble_preds = (pred1 + pred2 + pred3) / 3.0
final_labels = np.argmax(ensemble_preds, axis=1)

# --- Task 3: Compare Accuracies ---

from sklearn.metrics import accuracy_score

acc1 = accuracy_score(test_labels, np.argmax(pred1, axis=1))
acc2 = accuracy_score(test_labels, np.argmax(pred2, axis=1))
acc3 = accuracy_score(test_labels, np.argmax(pred3, axis=1))
ensemble_acc = accuracy_score(test_labels, final_labels)

print(f"\nModel 1 Accuracy: {acc1:.4f}")
print(f"Model 2 Accuracy: {acc2:.4f}")
print(f"Model 3 Accuracy: {acc3:.4f}")
print(f"ENSEMBLE ACCURACY: {ensemble_acc:.4f}")
```
